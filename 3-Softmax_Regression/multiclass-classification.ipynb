# Multi-class Classification Problem

In the previous problem, I have learnt about Logistic Regression algorithm for classification problem, with the goal is seperating two states (yes or no). And today, we will learn about an algorithm for classifying more than 2 classes, it is called Softmax Regression, or Cross-entropy minization.

Suppose that $k$ be the number of classes and $C$ be the set of classes ($\# C = k$). Then for each class $c \in C$, the goal is minimizing the Cross-entropy function: $$\text{Cross-entropy}(p, \hat{p}) = -\log\left(\frac{e^{\theta _c^Tx}}{\sum _{c \in C}e^{\theta _c^T.x}}\right).$$ And we will use Gradient descent technique for this function.

This is the raw idea for to solve this problem, I do not think we should implement from scratch it, because we have implemnted Gradient descent in the previous lesson. So now, I only show how to use model from Scikit-learn library for this problem.

```python
import kagglehub
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Load dataset
path = kagglehub.dataset_download("uciml/iris")
filename = f"{path}/{os.listdir(path)[0]}"

df = pd.read_csv(filename, index_col = 'Id')
print(df.head())

species_mapping = {
    'Iris-setosa': 1,
    'Iris-versicolor': 2,
    'Iris-virginica': 3
}

df['Species_encoded'] = df['Species'].map(species_mapping)

X = df.drop(['Species', 'Species_encoded'], axis = 1).values.astype(np.float64)
y = df['Species_encoded'].values.astype(np.float64)

features = X.shape[1]
classes = 3

print("Dataset shape:", X.shape)
print("Classes:", np.unique(y))

# 2. Split Train, Validation, and Test
val_size = 0.2
test_size = 0.125
random_state = 2
is_shuffle = True

X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size = val_size,
    random_state = random_state,
    shuffle = is_shuffle
)

X_train, X_test, y_train, y_test = train_test_split(
    X_train, y_train,
    test_size = test_size,
    random_state = random_state,
    shuffle = is_shuffle
)

# 3. Scale data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Train model
model = LogisticRegression(
    solver = 'lbfgs',
    max_iter = 1000,
    random_state = 42,
    verbose = 1
)

model.fit(X_train_scaled, y_train)
theta = np.concatenate((model.intercept_.reshape((1, classes)), model.coef_.reshape((features, classes))), axis = 0)

theta_1 = theta[:, 0]
theta_2 = theta[:, 1]
theta_3 = theta[:, 2]

print(f"theta of class 1: {theta_1}")
print(f"theta of class 2: {theta_2}")
print(f"theta of class 3: {theta_3}")

# 4. Evaluate
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.4f} %")

# 5. Predict
sample = np.array([[5, 3, 1.4, 0.2], [1, 1, 1, 1]])

sample = scaler.transform(sample)
y_sample_pred = model.predict(sample)

print(y_sample_pred)
```
