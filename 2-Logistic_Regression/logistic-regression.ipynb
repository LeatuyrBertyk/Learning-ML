# Logistic Regression

Logistic regression is used for classification when we want $h_{\theta}\in[0;1]$. We define $$h_{\theta}(x)=g(\theta^Tx)=\frac{1}{1+e^{-\theta^Tx}},$$ where $g(z)=\frac{1}{1+e^{-x}}$ is callded "sigmoid" or "logistic" function.

We have $P(y=1|x;\theta)=h_{\theta}(x)$ and $P(y=0|x;\theta)=1-h_{\theta}(x)$. We can merge those formula by $P(y|x;\theta)=h_{\theta}(x)^y\cdot \left(1-h_{\theta}(x)\right)^{1-y}$. Assume that the training set is an IID, then the **likelihood** function $L(\theta)$ is defined by $$L(\theta)=P(\vec{y}|x;\theta)=\prod_{i=1}^{m}P(y^{(i)}|x^{(i)};\theta)=\prod_{i=1}^mh_\theta(x^{(i)})^{y^{(i)}}\left(1-h_\theta(x^{(i)})\right)^{1-y^{(i)}}.$$

Then the **log-likelihood** function $l(\theta)$ is defined by $$l(\theta)=\log(L(\theta))=\sum_{i=1}^m\left[y^{(i)}\log\left(h_\theta(x^{(i)})\right)+(1-y^{(i)})\log\left(1-h_\theta(x^{(i)})\right)\right].$$

_Note:_ We easily compute that $l'(\theta)=\sum_{i=1}^m\left(y^{(i)}-h_\theta(x^{(i)})\right)x_j^{(i)}$. You can see that this formula is similar to the derivative of cost function in Linear regression.

Now, based on MLE (Maximum likelihood estimation), we have to maximize $l(\theta)$. To do this, we will apply the opposite algorihtm of Gradient Descent is **Gradient Ascent**. For detail, in each iter:

- with Batch Gradient Ascent, $\theta := \theta+\alpha\cdot\sum _{i=1}^m\left(y^{(i)}-h_\theta(x^{(i)})\right)x^{(i)}$;
- with Stochastic Gradient Ascent, $\theta := \theta+\alpha\cdot\left(y^{(i)}-h_\theta(x^{(i)})\right)x^{(i)}$.

Below is my implementation to demonstrate how the **Stochastic Gradient Ascent** works for optimizing logistic function.

```python
import numpy as np

def compute_sigmoid(theta, xi):
    z = np.dot(theta, xi)
    return 1 / (1 + np.exp(-z))

def stochastic_logistic_regression(X, Y, epochs = 10000, learning_rate = 1e-5):
    (m, n) = X.shape
    theta = np.zeros(n)

    for _ in range(epochs):
        loss = 0
        for i in range(m):
            prediction = compute_sigmoid(theta, X[i])
            error = Y[i] - prediction
            theta = theta + learning_rate * error * X[i]

    return theta
```
