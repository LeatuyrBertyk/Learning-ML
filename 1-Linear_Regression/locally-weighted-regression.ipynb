# Locally weighted regression

There are two important definition:

- "Parametric" learning algorithm: Fit fixed set of parameters to data.
- "Non-parametric" learning algorithm: Amount of data/parameters you nned to keep grow (linearly) with isze of data.

Locally weighted regression is a "non-parametric" learning algorithm. This model is based on its ancestor - Linear regression, but has some quite differences. We also minimize the cost function when implement this, but the cost function of this model has some new features: For each input $x$, we have to minimize $$L(\theta)=\sum_{i=1}^m\omega^{(i)}\left(y^{(i)}-\theta^Tx^{(i)}\right)^2$$ where $w^{(i)}$ is defined by $w^{(i)}=\exp\left(-\frac{\left(x^{(i)}-x\right)^2}{2\tau^2}\right)$ and $\tau$ is the bandwidth, which decides how many nearby examples to fit the line.

The definition of $w^{(i)}$ seems very weird, but it comes from the consequence of Gaussian Distribution.

Two common solutions for this problem is Gradient Descent or Normal Equation. Now, for each input $x$, I perform this optimization by **Gradient Descent** algorithm. Below is my implementation for **Stochastic Gradient Descent** method for this model.

```python
import numpy as np
import math

def compute_weight(xi, x, tau = 1):
    return math.exp(-(np.sum(xi - x))**2/(2*tau**2)) # Based on Gaussian Distribution

def stochastic_locally_weighted_regression(X, y, x, epochs = 10000, learning_rate = 1e-5, tau = 1):
    (m, n) = X.shape
    theta = np.zeros(n)
    losses = []

    for _ in range(epochs):
        loss = 0
        for i in range(m):
            wi = compute_weight(X[i], x, tau)
            prediction = np.dot(theta, X[i])
            error = prediction - y[i]
            theta = theta - learning_rate * wi * 2 * error * X[i]
            loss += wi * error**2
        losses.append(loss/m)

    return (np.dot(theta, x), losses)
```
