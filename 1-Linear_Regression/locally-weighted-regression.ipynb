# Locally weighted regression

```python
import numpy as np
import math

def compute_weight(xi, x):
    return math.exp(-(xi - x)**2/2) # Based on Gaussian Distribution

def stochastic_locally_weighted_regression(X, Y, x, epochs = 10000, learning_rate = 1e-5):
    (m, n) = X.shape
    theta = np.zeros(n)

    for _ in range(epochs):
        for i in range(m):
            wi = compute_weight(X[i], x)
            prediction = np.dot(theta, X[i])
            error = prediction - Y[i]
            theta = theta - learning_rate*wi*2*error*X[i]

    return theta
```
