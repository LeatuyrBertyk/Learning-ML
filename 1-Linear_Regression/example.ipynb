# Linear Regression

This is an example comparing **Normal Equation** method and model from **Scikit-learn** library to solve optimum for Linear Regression problem.

I used data from [Kaggle-Housing Price Dataset](https://www.kaggle.com/datasets/yasserh/housing-prices-dataset).

```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

path = kagglehub.dataset_download("yasserh/housing-prices-dataset")
df = pd.read_csv(f"{path}/Housing.csv")

print(f"Shape: {df.shape}")
print(df.head()) # First 5 lines of data frame

X = df[['area', 'bedrooms', 'bathrooms', 'stories', 'parking']].values
y = df['price'].values.reshape(-1, 1)

# 1. Normal equation:
m = X.shape[0]
X_bias = np.column_stack([np.ones(m), X])

theta_normal = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y

# 2. Scikit-learn model:
model = LinearRegression()
model.fit(X, y)

theta_sklearn = np.column_stack([model.intercept_,model.coef_]).reshape(-1,1)

# 3. Result:
print(f"Normal Equation:\n{theta_normal}")
print(f"Scikit-learn:\n{theta_sklearn}")
```
