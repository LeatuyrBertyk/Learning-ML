# Linear Regression

This is an example comparing **Normal Equation** method and model from **Scikit-learn** library to solve optimum for Linear Regression problem.

I used data from [Kaggle-Housing Price Dataset](https://www.kaggle.com/datasets/yasserh/housing-prices-dataset).

```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

path = kagglehub.dataset_download("yasserh/housing-prices-dataset")
df = pd.read_csv(f"{path}/Housing.csv")

print(f"Data frame shape: {df.shape}")
print(df.head(3)) # Fist 3 lines of data frame

m = df.shape[0]

X = df[['area', 'bedrooms', 'bathrooms', 'stories']].values
Y = df['price'].values.reshape(-1, 1)

# 1. Normal equation:
X_base = np.concatenate((np.ones((m,1)), X), axis = 1) # Add a column filled by 1 at the left of data frame
theta_normal = np.linalg.inv(X_base.T @ X_base) @ X_base.T @ Y

# 2. Scikit-learn:
model = LinearRegression()
model.fit(X, Y)

theta_sklearn = np.concatenate((model.intercept_.reshape(-1,1),model.coef_.reshape(-1,1)), axis = 0)

# 3. Result:
print(f"Normal Equation:\n{theta_normal}")
print(f"Scikit-learn:\n{theta_sklearn}")
```
