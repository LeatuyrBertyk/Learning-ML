# Linear Regression

This is my implementation to introduce an mini-project using Linear Regression in housing price prediction. I used training data from [Kaggle-Housing Price Dataset](https://www.kaggle.com/datasets/yasserh/housing-prices-dataset).

For detail, I proposed a linear function to predict housing price with 3 features: area, number of bedrooms, and number of bathrooms. 

To have a deep understand about this machine learning model, I implemented **Gradient Descent** method by Stochastic algorithm to get optimal function, and determined root of **Normal Equation** method (as small dataset). At the end, I checked those results with given model in **Scikit-learn** library. 

First, I implemented **Normal Equation** and model from **Scikit-learn**:

```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Prepare dataset:
path = kagglehub.dataset_download("yasserh/housing-prices-dataset")
df = pd.read_csv(f"{path}/Housing.csv")

print(f"Data frame shape: {df.shape}")
print(df.head(3)) # Fist 3 lines of data frame to check

m = df.shape[0] # Total training datas

X = df[['area', 'bedrooms', 'bathrooms']].values
Y = df['price'].values

# 2. Normal equation:
X_bias = np.concatenate((np.ones((m,1)), X), axis = 1) # Add a column filled by 1 at the left of data frame
theta_normal = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ Y

# 3. Scikit-learn:
model = LinearRegression()
model.fit(X, Y)

theta_sklearn = np.concatenate((model.intercept_.reshape(-1,1),model.coef_.reshape(-1,1)), axis = 0)

# 4. Result:
print(f"Normal Equation:\n{theta_normal.reshape(-1, 1)}")
print(f"Scikit-learn:\n{theta_sklearn}")
```

As this is a small dataset, **Normal Equation** can compute the exact root promptly in spite of its disadvantages about time complexity. We can see that, those two model give the same result.

With the harder, I implemented **Gradient Descent** method below:

```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

def stochastic_linear_regression(epochs, learning_rate, X, Y):
    (m, n) = X.shape
    theta = np.zeros(n)
    losses = []

    for _ in range(epochs):
        total_loss = 0
        for i in range(0, m):
            prediction = np.dot(theta, X[i])
            error = prediction - Y[i, 0]
            theta = theta - learning_rate*error*X[i]

            total_loss += error**2
        losses.append(total_loss/m)

    return (theta, losses)

# 1. Prepare dataset:
path = kagglehub.dataset_download("yasserh/housing-prices-dataset")
df = pd.read_csv(f"{path}/Housing.csv")

print(f"Data frame shape: {df.shape}")
m = df.shape[0]

X = df[['area', 'bedrooms', 'bathrooms']].values
Y = df['price'].values.reshape(-1, 1)

# 2. Normalize dataframe due to its large scale:
#scaler = StandardScaler()
#X = scaler.fit_transform(X)
#Y = scaler.fit_transform(Y)

X_bias = np.concatenate((np.ones((m, 1)), X), axis = 1)

# 3. Result:
(theta, losses) = stochastic_linear_regression(10000, 1e-5, X_bias, Y)
theta = theta.reshape(-1, 1)
print(theta)
```
