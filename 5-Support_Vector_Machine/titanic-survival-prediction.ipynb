# Support Vector Machine

With classification problem, we have learnt two main algorithm, that is Logistic Regression and Gaussian Discriminant Analysis, which draw a straight line to seperate.

But for some cases, we may need a non-linear boudary, and Support Vector Machine algorithm with Kernel trick will help us. 

- **Step 1:** Choose a suitable Kernel function $K(x,z)$ in four common ones:

  - Linear: $K(x,z)=x^Tz$
  - Sigmoid: $K(x,z) = \tanh (\gamma x^Tz+r)$
  - Polynomial: $K(x,z)=(\gamma x^Tz+r)^d$
  - Radial Basis Function (Gaussian Kernel): $K(x,z)=\exp (-\gamma \| x-z \|^2), \gamma >0$ (the most common)
- **Step 2:** We obtain the optimization: $$\max _{\alpha} W(\alpha)=\sum _{i=1}^m \alpha _i -\frac{1}{2} \sum _{i=1}^m y^{(i)} y^{(j)} \alpha _i \alpha _j K\left(x^{(i)}, x^{(j)}\right)$$ $$\text{s.t. } 0\leq \alpha _i \leq C, i=1, 2,\ldots, m \text{ and } \sum _{i=1}^m\alpha _iy^{(i)}=0$$
which also be called "Dual optimization problem". 
- **Step 3:** We already get $\alpha$, now the prediction can be made via $$h_{\omega, b} (x)=g(\omega^Tx+b)=g\left(\sum _{i=1}^m \alpha _i y^{(i)} K\left( x^{(i)}, x \right)+b\right)$$

  where $$g(x) = \begin{cases}
    +1, & \text{if } x \geq 0 \\
    -1, & \text{otherwise}
  \end{cases}$$

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC

# ===== Preprocessing =====

# Load dataset
df = pd.read_csv('titanic_modified_dataset.csv', index_col = 'PassengerId') # Choose PassengerID as index

df = df.to_numpy().astype(np.float64)

X = df[:, :-1]
y = df[:, -1]

# Split Train and Test
test_size = 0.3
random_state = 2
is_shuffle = True

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = test_size,
    random_state = random_state,
    shuffle = is_shuffle
)

(m, n) = X_train.shape

# Scale data
normalizer = StandardScaler()
X_train[:, 1:] = normalizer.fit_transform(X_train[:, 1:])
X_test[:, 1:] = normalizer.transform(X_test[:, 1:])

# ===== SVM =====

kernels = {
    'Linear': {'kernel': 'linear', 'gamma': 1/m, 'C': 1000},
    'Sigmoid': {'kernel': 'sigmoid', 'gamma': 1/m, 'C': 1000},
    'Polynomial': {'kernel': 'poly', 'degree': 2, 'gamma': 1/m, 'C': 1000},
    'RBF': {'kernel': 'rbf', 'gamma': 1/m, 'C': 1000}
}

for name, params in kernels.items():
    clf = SVC(**params)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    print(f"Accuracy {name}: {accuracy_score(y_test, y_pred)*100:.4f}%")
```
